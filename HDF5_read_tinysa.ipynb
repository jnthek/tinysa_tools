```python
import matplotlib.pyplot as plt
import numpy as np
import h5py
import time

plt.rcParams.update({'font.size': 18})
```

```python
# h5_file = "20261010-121137_tinysa.h5"
h5_file = "20261010-122045_tinysa.h5"
# h5_file = "20261010-134156_tinysa.h5"

print ("Reading",h5_file)
with h5py.File(h5_file, 'r') as hf:
    f_low = hf["data"].attrs['f_low']
    f_high = hf["data"].attrs['f_high']
    RBW = hf["data"].attrs['RBW']
    N_points = hf["data"].attrs['N_points']

    timestamps = hf["data/timestamps"][()][1:]
    radio = hf["data/radio"][()][1:]
```

```python
freq = np.linspace(f_low, f_high, N_points)

plt.figure(figsize=(16,9))
plt.imshow(radio, aspect='auto', cmap='jet')
```

```python
radio.shape
plt.figure(figsize=(16,9))
plt.plot(freq/1e6, 10*np.log10(np.mean(10**(radio[:,:]/10), axis=0)))
```
